```python
import pandas as pd
import numpy as np
from scipy import stats
from pathlib import Path

# Read data
DATA_DIR = Path.cwd() / "testfiles_" / "data"
CSV_PATH = DATA_DIR / "test7_1.csv"
df = pd.read_csv(CSV_PATH, header=0)

# Calculate returns
returns = df.iloc[:, 0].dropna()

# Calculate mean and standard deviation
mean_return = returns.mean()
std_return = returns.std()

# Set confidence level
confidence_level = 0.95

# Calculate VaR using Normal Distribution
z_score = stats.norm.ppf(1 - confidence_level)
var_diff_from_mean = -z_score * std_return
var_absolute = -(mean_return + z_score * std_return)

# Print results
print(f"VaR Absolute: {var_absolute}") # actual VaR value (the return level at the confidence threshold)
print(f"VaR Diff from Mean: {var_diff_from_mean}") # the distance from the mean to the VaR threshold
```

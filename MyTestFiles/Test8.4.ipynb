```python
import pandas as pd
import numpy as np
from scipy import stats
from pathlib import Path

# Read data
DATA_DIR = Path.cwd() / "testfiles_" / "data"
CSV_PATH = DATA_DIR / "test7_1.csv"
df = pd.read_csv(CSV_PATH, header=0)

# Use the first numeric column as returns
ret = df.select_dtypes(include=[float, int]).iloc[:, 0].dropna().astype(float)
mu = float(ret.mean())
sigma = float(ret.std(ddof=1))
if not np.isfinite(sigma) or sigma <= 0:
    raise ValueError("Invalid std for ES computation.")

# ES from Normal (left tail) at 95%
alpha = 1.0 - 0.95
z = stats.norm.ppf(alpha)
phi = stats.norm.pdf(z)
es_diff = sigma * phi / alpha
es_abs = -(mu - sigma * phi / alpha)     # absolute ES level
                # distance from mean
# Print exactly (no CSV file)
print("ES Absolute,ES Diff from Mean")
print(f"{es_abs:.12f},{es_diff:.12f}")
```

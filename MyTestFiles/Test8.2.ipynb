```python
import pandas as pd
import numpy as np
from scipy import stats
from pathlib import Path

# Read data
DATA_DIR = Path.cwd() / "testfiles_" / "data"
CSV_PATH = DATA_DIR / "test7_2.csv"
df = pd.read_csv(CSV_PATH, header=0)

# Calculate returns
returns = df.iloc[:, 0].dropna()

# Fit T distribution to the data
df_t, loc, scale = stats.t.fit(returns)

# Set confidence level
confidence_level = 0.95

# Calculate VaR using fitted T Distribution
t_score = stats.t.ppf(1 - confidence_level, df_t)
var_diff_from_mean = - t_score * scale
var_absolute = - (loc +  t_score * scale)

# Print results
print(f"VaR Absolute: {var_absolute}") # actual VaR value (the return level at the confidence threshold)
print(f"VaR Diff from Mean: {var_diff_from_mean}") # the distance from the mean to the VaR threshold
```

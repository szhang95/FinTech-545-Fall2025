```python
import numpy as np
import pandas as pd
from pathlib import Path

# Settings
N_SIMS = 100_000
SEED = 42

# Read data
DATA_DIR = Path.cwd() / "testfiles_" / "data"
CSV_PATH = DATA_DIR / "test5_3.csv"

# Read covariance (first row has column names)
df = pd.read_csv(CSV_PATH, header=0)
cols = list(df.columns)
Sigma = df.to_numpy(dtype=float)

# Symmetrize
Sigma = (Sigma + Sigma.T) / 2.0

# near_psd fix (eigenvalue clipping)
d, V = np.linalg.eigh(Sigma)
d = np.clip(d, 0.0, None)
B = V * np.sqrt(d)  # Sigma_fixed = B B^T

# Simulate X ~ N(0, Sigma_fixed)
rng = np.random.default_rng(SEED)
n = Sigma.shape[0]
Z = rng.standard_normal(size=(N_SIMS, n))
X = Z @ B.T

# Sample covariance (output)
S_hat = np.cov(X, rowvar=False, ddof=1)

print(", ".join(cols))
for i in range(n):
    row = ", ".join(f"{S_hat[i, j]:.16f}" for j in range(n))
    print(row)
```

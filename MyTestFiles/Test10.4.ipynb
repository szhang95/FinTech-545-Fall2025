```python
import pandas as pd
import numpy as np
from pathlib import Path
from scipy.optimize import minimize

DATA_DIR = Path.cwd() / "testfiles_" / "data"

cov_df = pd.read_csv(DATA_DIR / "test5_2.csv", header=0)
means_df = pd.read_csv(DATA_DIR / "test10_3_means.csv", header=0)

cov_matrix = cov_df.values
mean_returns = means_df['Mean'].values

rf = 0.04

def negative_sharpe_ratio(weights, mean_returns, cov_matrix, rf):
    portfolio_return = np.dot(weights, mean_returns)
    portfolio_variance = np.dot(weights, np.dot(cov_matrix, weights))
    portfolio_std = np.sqrt(portfolio_variance)
    sharpe_ratio = (portfolio_return - rf) / portfolio_std
    return -sharpe_ratio

n_assets = len(mean_returns)

constraints = {'type': 'eq', 'fun': lambda x: np.sum(x) - 1}
bounds = tuple((0.1, 0.5) for _ in range(n_assets))
initial_weights = np.array([1/n_assets] * n_assets)

result = minimize(
    negative_sharpe_ratio,
    initial_weights,
    args=(mean_returns, cov_matrix, rf),
    method='SLSQP',
    bounds=bounds,
    constraints=constraints,
    options={'ftol': 1e-12, 'maxiter': 1000}
)

weights = result.x
weights = weights / np.sum(weights)

print("w")
for w in weights:
    print(f"{w:.17f}")
```

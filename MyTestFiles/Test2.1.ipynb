```python
import numpy as np
import pandas as pd
from pathlib import Path

LAMBDA = 0.97
DATA_DIR = Path.cwd() / "testfiles_" / "data"
df = pd.read_csv(DATA_DIR / "test2.csv")

# numeric data (listwise)
num = df.select_dtypes(include=[np.number]).dropna()
X = num.to_numpy(float)
n, d = X.shape
if n < 2 or d < 2:
    raise ValueError("Need at least 2 observations and 2 numeric columns.")

# normalized exponential weights (newest gets largest weight)
w = (1 - LAMBDA) * LAMBDA ** np.arange(n-1, -1, -1)
w = w / w.sum()                  # finite-sample normalization

# weighted mean and EW covariance
mu = (w[:, None] * X).sum(axis=0)
Xc = X - mu
S = (w[:, None] * Xc).T @ Xc     # d x d

ew_cov = pd.DataFrame(S, index=num.columns, columns=num.columns)
print(ew_cov)
```

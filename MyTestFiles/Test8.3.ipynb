```python
import pandas as pd
import numpy as np
from pathlib import Path
from scipy import stats

# Read data
DATA_DIR = Path.cwd() / "testfiles_" / "data"
CSV_PATH = DATA_DIR / "test7_2.csv"
df = pd.read_csv(CSV_PATH, header=0)

# Calculate returns
returns = df.iloc[:, 0].dropna()

# Fit T distribution to the data
df_t, loc, scale = stats.t.fit(returns)
# loc → location → mean → return
# scale → spread → std → sigma (σ)

# Set confidence level
confidence_level = 0.95

# Number of simulations
n_simulations = 10000

# Generate simulated returns from fitted T distribution
np.random.seed(32)
simulated_returns = stats.t.rvs(df_t, loc, scale, size=n_simulations)

# Calculate VaR from simulated data
var_absolute = - np.percentile(simulated_returns, (1 - confidence_level) * 100)
var_diff_from_mean = loc - np.percentile(simulated_returns, (1 - confidence_level) * 100)

# Print results
print(f"VaR Absolute: {var_absolute}") # actual VaR value (the return level at the confidence threshold)
print(f"VaR Diff from Mean: {var_diff_from_mean}") # the distance from the mean to the VaR threshold
```

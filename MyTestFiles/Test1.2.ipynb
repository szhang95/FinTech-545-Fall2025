```python
import numpy as np
import pandas as pd
from pathlib import Path

# Read data and give output
DATA_DIR = Path.cwd() / "testfiles_" / "data"
df = pd.read_csv(DATA_DIR / "test1.csv")

# drop any row with missing values)
df_clean = df.dropna(axis=0, how="any")

# Use numeric columns only
num_df = df_clean.select_dtypes(include=[np.number])
if num_df.shape[1] < 2:
    raise ValueError("Need at least 2 numeric columns to compute a covariance matrix.")

corr_mat = df_clean.corr(method="pearson")
#print("Rows before:", len(df), "| after dropna:", len(df_clean))
print(corr_mat.to_string())
```

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Read data and give output
DATA_DIR = Path.cwd() / "testfiles_" / "data"
FILENAME = "test6.csv"

df = pd.read_csv(DATA_DIR / FILENAME)
date = pd.to_datetime(df.iloc[:, 0], errors="coerce")
prices = df.iloc[:, 1:].apply(pd.to_numeric, errors="coerce")

# Manual arithmetic returns: r_t = P_t / P_{t-1} - 1
P = prices.to_numpy(dtype=float)
if P.shape[0] < 2:
    raise ValueError("Need at least 2 rows to compute returns.")
R = P[1:, :] / P[:-1, :] - 1.0

# Build output with Date column aligned to t (the later row)
out = pd.DataFrame(R, columns=prices.columns)
out.insert(0, "Date", date.iloc[1:].reset_index(drop=True))
out
```

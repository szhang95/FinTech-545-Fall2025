```python
import numpy as np
import pandas as pd
from pathlib import Path

# Load: first column is Date
DATA_DIR = Path.cwd() / "testfiles_" / "data"
df = pd.read_csv(DATA_DIR / "test6.csv")
date = pd.to_datetime(df.iloc[:, 0], errors="coerce")
prices = df.iloc[:, 1:].apply(pd.to_numeric, errors="coerce")

# Log returns: ln(P_t / P_{t-1}) = ln(P_t) - ln(P_{t-1})
logP = np.log(prices.to_numpy(dtype=float))
R = logP[1:, :] - logP[:-1, :]

out = pd.DataFrame(R, columns=prices.columns)
out.insert(0, "Date", date.iloc[1:].reset_index(drop=True))
out
```
